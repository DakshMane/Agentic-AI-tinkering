```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from dataclasses import dataclass

@dataclass
class ColorContext : 
    favorite_color : str = "blue"
    least_favorite_color : str = "yellow"

```

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

model = init_chat_model(model="gemini-2.5-flash" , model_provider="google-genai")
agent = create_agent(model=model ,context_schema=ColorContext)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke({
    "messages" : [HumanMessage(content="What is my favorite color ? ")]
 } , context=ColorContext()
 )
```

```python
from pprint import pprint
pprint(response)
```

```python
from langchain.tools import tool , ToolRuntime

@tool 
def get_favorite_color(runtime : ToolRuntime) -> str : 
    """Get favorite color of user """

    return runtime.context.favorite_color

@tool 
def get_least_favorite_color(runtime : ToolRuntime) -> str : 
    """Get least  favorite color of user """

    return runtime.context.least_favorite_color 
```

```python
agent = create_agent(
    model= model , 
    tools=[get_favorite_color , get_least_favorite_color] , 
    context_schema=ColorContext
)
```

```python
response = agent.invoke({
    "messages" : [HumanMessage(content="What is my favorite color ? ")]
    
} , 

context=ColorContext())

pprint(response)
```

```python

response = agent.invoke(
    {"messages": [HumanMessage(content="What is my favourite colour?")]},
    context=ColorContext(favorite_color="green")
)

pprint(response)
```
