```python
from dotenv import load_dotenv
load_dotenv()
from langchain.chat_models import init_chat_model

model = init_chat_model(model="gemini-2.5-flash" , model_provider="google_genai" ,
                        temperature= 1.0 
                        )

response = model.invoke("What is capital of Moon ")

print(response.content)
```

```python
from langchain.agents import create_agent 

from langchain_core.messages import HumanMessage , AIMessage
from pprint import pprint

agent = create_agent(model=model , system_prompt="You are a expert helper assistant")


response = agent.invoke({
    "messages" : [HumanMessage(content="What is capital of Moon ? ")]
})

pprint(response['messages'][-1].content)
```

```python
from langchain.agents import create_agent 

from langchain_core.messages import HumanMessage , AIMessage
from pprint import pprint
agent = create_agent(model=model , system_prompt="You are a expert helper assistant")

response = agent.invoke({
    "messages" : [HumanMessage(content="What is capital of Moon ? ") , 
                  AIMessage(content="The capital of Moon is Luna ")  , 
                  HumanMessage(content="Interesting , tell me more about Luna ")] 
})

pprint(response['messages'][-1].content)
```

## STREAMING OUTPUTS  ....

```python
from langchain.agents import create_agent 

from langchain_core.messages import HumanMessage , AIMessage
from pprint import pprint
agent = create_agent(model=model , system_prompt="You are a expert helper assistant")


for token , metadata in agent.stream({
    "messages" : [HumanMessage(content="Tell me all about Luna City, the capital of the Moon")]
} , stream_mode="messages") :

     if token.content:  # Check if there's actual content
        print(token.content, end="", flush=True)  # Print token
```
