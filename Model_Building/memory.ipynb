```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from langchain.chat_models import init_chat_model
model = init_chat_model(model="gemini-2.5-flash" , model_provider="google_genai" ,
                        temperature= 1.0 
                        )


```

```python
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    model=model , 
    checkpointer=InMemorySaver() , 
   
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Daksh , my favorite color is green")
config = {"configurable" : {"thread_id" : "1"}}

response = agent.invoke(
    {"messages" : [question]} , config
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="Do you remember what is my favorite color and my name ? ")

response = agent.invoke(
    {
        "messages" : [question] 
    } , config 
)

pprint(response)
```

```python
question = HumanMessage(content="Actually my full name is Hello World Daksh Mane , my color changed to blue ")

response = agent.invoke({"messages" : [question]} , config)

```

```python
question = HumanMessage(content="Do you remember my old name and my old favorite color , im just testing you")
response = agent.invoke({"messages" : [question]} , config)

pprint(response)
```
