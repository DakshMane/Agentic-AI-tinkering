```python
from dotenv import load_dotenv 
load_dotenv()
```

```python
from langchain.tools import tool 


@tool
def square_root(x : float ) -> float :
    """Calculate the square root of a number"""
    return x**0.5
```

```python
@tool("square_root")
def tool1(x : float) -> float :
    """Calculate square root of number"""
    return x**0.5
```

```python
@tool("square_root" , description="Calculate square root of number")
def tool1(x:float) -> float : 
    return x**0.5
```

```python
tool1.invoke({"x" : 467})
```

# ADDING TO AGENTS 

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
model = init_chat_model(model="gemini-2.5-flash" , model_provider="google_genai" ,
                        temperature= 1.0 
                        )

agent = create_agent(
    model=model  , 
    tools=[tool1] , 
    system_prompt="You are an arithmetic wizard . Use tools to calculate the square root and square of any number"
)

```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="What is square root of 225")

response = agent.invoke({
    'messages' : [question]
})

print(response['messages'][-1].content)
```
